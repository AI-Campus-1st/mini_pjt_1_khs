```python
ageg_df['region_name'] = (
    ageg_df['행정구역']
    .str.replace('경기도','',regex=False)
    .str.replace(r'\([^)]*\)', '',regex=True)
    .str.strip() #공백때문에 strip 추가
)
```

```python
region_names = ageg_df['region_name'].unique() #소계만 추리기

del_regions = []

for region in region_names:
    sub_region = any(
        name.startswith(region + ' ')
        for name in region_names
        if name != region
    )
    
    if sub_region:
        del_regions.append(region)
```

```python
sorted(del_regions) #strip 추가안했을땐 실행안되고 추가후에 실행됨
```

```python
ageg_df = ageg_df[ #상위 시(소계) 삭제 코드
    ~ageg_df['region_name'].isin(del_regions)
].copy()
```

```python
ageg_df = ageg_df[ #경기도(총계) 삭제 코드
    ageg_df['region_name'].str.strip() != ''
].copy()
```

```python
base_regions = set(store_df['시군구명'].unique()) #전처리 확인
age_regions = set(ageg_df['region_name'].unique())

print('상권dt 지역 수', len(base_regions))
print('인구db 지역 수', len(age_regions))
print('공통 지역 수', len(base_regions & age_regions))
print('상권에만 있는 지역', sorted(base_regions - age_regions))
print('인구에만 있는 지역', sorted(age_regions - base_regions))
```
